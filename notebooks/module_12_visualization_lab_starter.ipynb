## Part 9: Correlation Heatmap

## Task: Create a correlation heatmap for the numeric columns in the cleaned penguins dataset

```python
# TODO: Fill in the numeric column names
numeric_cols=[

    "bill_length_mm",
    "bill_depth_mm",
    "flipper_length_mm",
    "body_mass_g"
    
]

# TODO: Calculate the correlation matrix

penguins_corr = penguins_clean[numeric_cols].corr()

penguins_corr



```

```python

# TODO: Create the heatmap

plt.figure(figsize=(10,6))

sns.heatmap(penguins_corr, annot=True,cmap="coolwarm",center=0)
plt.title("Correlation Heatmap of Numeric Penguin anatomy Variables")
plt.show()
```
